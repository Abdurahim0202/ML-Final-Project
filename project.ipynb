```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('state_NJ.csv')
data
```

```python
data.columns
```

```python
data = data.drop(data[["activity_year", "lei", "state_code", "census_tract"]], axis=1)
data
```

```python
data = data.drop(data[["purchaser_type", "denial_reason-1", "denial_reason-2", "denial_reason-3", "denial_reason-4"]], axis=1)
data
```

```python
data = data.drop(data[["aus-3", "aus-4", "aus-5"]], axis=1)
data
```

We have Derived Ethnicity which is the combination of applicants and co-applicants ethnicity. Additionally for each applicant we have 5 features showing their ethnicity, I want to examine if we can safely reduce the number of features

```python
for i in range(1, 6):
    col = f"applicant_ethnicity-{i}"

    print(f"\nEthnicity #{i} distribution:")
    print(data[col].value_counts())

    missing_count = data[col].isnull().sum()
    missing_percentage = data[col].isnull().mean() * 100

    print(f"Ethnicity #{i} missing values: {missing_count}")
    print(f"Ethnicity #{i} missing percentage: {missing_percentage:.2f}%")
```

```python
data = data.drop(data[["applicant_ethnicity-2", "applicant_ethnicity-3", "applicant_ethnicity-4", "applicant_ethnicity-5"]], axis=1)
data
```

```python
for i in range(1, 6):
    col = f"co-applicant_ethnicity-{i}"

    print(f"\nEthnicity #{i} distribution:")
    print(data[col].value_counts())

    missing_count = data[col].isnull().sum()
    missing_percentage = data[col].isnull().mean() * 100

    print(f"Ethnicity #{i} missing values: {missing_count}")
    print(f"Ethnicity #{i} missing percentage: {missing_percentage:.2f}%")
```

```python
data = data.drop(data[["co-applicant_ethnicity-2", "co-applicant_ethnicity-3", "co-applicant_ethnicity-4", "co-applicant_ethnicity-5"]], axis=1)
data
```
